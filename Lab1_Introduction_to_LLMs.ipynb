```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

sentences = [
    "I am learning Generative AI.",
    "Hello world!",
    "Tokenization is interesting.",
    "ChatGPT is useful for students."
]

for sentence in sentences:
    tokens = tokenizer.tokenize(sentence)
    token_ids = tokenizer.convert_tokens_to_ids(tokens)

    print("Sentence:", sentence)
    print("Tokens:", tokens)
    print("Token IDs:", token_ids)
    print("Number of tokens:", len(tokens))
    print("-" * 40)
```

### Observations and Answers

1. **Is one word always one token?** No, a word can contain one or multiple tokens.
2. **Why are some words split?** The tokenizer uses a fixed vocabulary and splits unfamiliar words into smaller parts.
3. **What happens to punctuation?** Punctuation can be represented as separate tokens.
4. **Why does the model need token IDs?** Token IDs convert text into numbers that the model can process.

**Conclusion:** I learned that tokenization converts text into tokens and their corresponding numerical IDs.

```python
!pip -q install transformers accelerate

import torch
from transformers import pipeline

print("GPU available:", torch.cuda.is_available())

pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device=0 if torch.cuda.is_available() else -1
)

prompt = "Explain artificial intelligence in simple words."
result = pipe(prompt, max_new_tokens=60, do_sample=False)

print(result[0]["generated_text"])
if torch.cuda.is_available():
    print("GPU memory used (MB):",
          round(torch.cuda.memory_allocated() / 1024**2, 2))
```

```python
models = [
    ("4B FP16", 4, 16),
    ("4B INT4", 4, 4),
    ("7B INT4", 7, 4)
]

print("Estimated model weight memory:")
for name, params, bits in models:
    memory = params * bits / 8
    print(name, ":", memory, "GB (approximately)")
```

### Activity 4: Can My Computer Run an LLM?

```python
from getpass import getpass
from groq import Groq

api_key = getpass("Enter your Groq API key: ")
client = Groq(api_key=api_key)

for temp in [0.1, 0.7, 1.0]:
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": "Suggest a creative project idea for college students."
            }
        ],
        temperature=temp,
        max_completion_tokens=120
    )

    print(f"\nTemperature: {temp}")
    print(response.choices[0].message.content)
    print("-" * 40)
```

### Activity 8: Temperature Experiment

**Observations and Answers**

**Conclusion:** I learned that temperature affects the variety and predictability of generated text.

```python
from groq import Groq
from getpass import getpass

api_key = getpass("Enter your Groq API key: ")
client = Groq(api_key=api_key)

messages = []

while True:
    user_input = input("You (type exit to stop): ")

    if user_input.lower() == "exit":
        print("Chat ended.")
        break

    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=messages,
        max_completion_tokens=150
    )

    reply = response.choices[0].message.content
    print("Bot:", reply)

    messages.append({"role": "assistant", "content": reply})
```

**1. Why does the chatbot need to remember previous messages?**
Ans : It helps the chatbot remember previous messages and give relevant answers.

**2. What happens if you drop the conversation history?**
Ans : he chatbot may not remember earlier messages and will treat each question more independently.

**3. What's the difference between this and a single one-off prompt?**
Ans :A single prompt gives one response, while a chatbot supports multiple connected messages.



```python
from groq import Groq
from getpass import getpass

api_key = getpass("Enter your Groq API key: ")
client = Groq(api_key=api_key)

messages = [
    {
        "role": "system",
        "content": "You are a friendly AI and ML tutor. Explain concepts in simple language with examples."
    }
]

while True:
    user_input = input("You (type exit to stop): ")

    if user_input.lower() == "exit":
        print("Chat ended.")
        break

    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=messages,
        max_completion_tokens=200
    )

    reply = response.choices[0].message.content
    print("AI/ML Tutor:", reply)

    messages.append({"role": "assistant", "content": reply})
```

**1. What is a system prompt?**

A system prompt gives the chatbot instructions about its role, behavior, and response style.

**2. How does changing the system prompt affect responses?**

It changes how the chatbot responds. For example, a tutor explains concepts simply, while a coding assistant focuses on programming.

**3. Give another example of a system prompt.**

“You are a coding assistant. Help users understand programming concepts and debug code with clear explanations.”

Conclusion: System prompts help customize a chatbot for different tasks without changing the underlying model.
